```python
## Q1. Arithmetic and Assignment Operators
```

```python
# Initializing two numeric variables
a = 15
b = 4

# --- Arithmetic Operations ---
print("--- Arithmetic Operations ---")
print("Addition (a + b):", a + b)
print("Subtraction (a - b):", a - b)
print("Multiplication (a * b):", a * b)
print("Division (a / b):", a / b)
print("Floor Division (a // b):", a // b)
print("Modulus (a % b):", a % b)
print("Exponentiation (a ** b):", a ** b)

# --- Assignment Operators ---
print("\n--- Assignment Operations on Variable 'a' ---")
print("Original value of a:", a)

a += 5  # Equivalent to a = a + 5
print("After a += 5:", a)

a -= 3  # Equivalent to a = a - 3
print("After a -= 3:", a)

a *= 2  # Equivalent to a = a * 2
print("After a *= 2:", a)

a /= 4  # Equivalent to a = a / 4
print("After a /= 4:", a)
```

```python
## Q2. Comparison and Logical Operators
```

```python
# Taking two integer inputs from the user
num1 = int(input("Enter first integer: "))
num2 = int(input("Enter second integer: "))

# --- Comparison Operators ---
print("\n--- Comparison Operators Results ---")
print(f"{num1} == {num2}:", num1 == num2)
print(f"{num1} != {num2}:", num1 != num2)
print(f"{num1} > {num2}:", num1 > num2)
print(f"{num1} < {num2}:", num1 < num2)
print(f"{num1} >= {num2}:", num1 >= num2)
print(f"{num1} <= {num2}:", num1 <= num2)

# --- Logical Operators ---
print("\n--- Logical Operators Results ---")
# Defining two condition blocks to combine
condition1 = (num1 > 0)
condition2 = (num2 > 0)

print(f"Condition 1 (num1 > 0): {condition1}")
print(f"Condition 2 (num2 > 0): {condition2}")
print("Condition 1 and Condition 2:", condition1 and condition2)
print("Condition 1 or Condition 2:", condition1 or condition2)
print("not Condition 1:", not condition1)
```

```python
## Q3. Bitwise Operators
```

```python
# Taking two integer inputs for bitwise calculations
x = int(input("Enter first integer: "))
y = int(input("Enter second integer: "))

# --- Bitwise Operations ---
print("\n--- Bitwise Operators Results ---")
print(f"Bitwise AND ({x} & {y}):", x & y)
print(f"Bitwise OR ({x} | {y}):", x | y)
print(f"Bitwise XOR ({x} ^ {y}):", x ^ y)
print(f"Bitwise NOT (~{x}):", ~x)
print(f"Left Shift ({x} << 1):", x << 1)
print(f"Right Shift ({x} >> 1):", x >> 1)
```
